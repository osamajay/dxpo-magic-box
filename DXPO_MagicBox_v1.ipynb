```python
# ============================================
# CELL 4: DATA LOADING & CLEANING
# ============================================

print("=" * 45)
print("🔧 DATA LOADING & CLEANING")
print("=" * 45)

# Load data
df = pd.read_excel('J-Bee-sample.xlsm',
                   sheet_name='OriginalData',
                   engine='openpyxl')

# Apply cleaning fixes
df['Music CDs-Qty'] = df['Music CDs-Qty'].fillna(0)
df['zip_cd'] = df['zip_cd'].astype(str).str[:5]

# Convert numeric columns
amt_cols = ['Book-Amt', 'Music CDs-Amt',
            'Housewares-Amt', 'Tools-Amt']
qty_cols = ['Book-Qty', 'Music CDs-Qty',
            'Housewares-Qty', 'Tools-Qty']
for col in amt_cols + qty_cols:
    df[col] = pd.to_numeric(df[col],
                            errors='coerce').fillna(0)

# Add calculated columns
df['Total-Amt'] = df[amt_cols].sum(axis=1)
df['Products-Bought'] = (df[amt_cols] > 0).sum(axis=1)

print(f"✅ Data loaded and cleaned!")
print(f"   Rows    : {len(df)}")
print(f"   Columns : {len(df.columns)}")
# Testing line by Jay
print(f"   Shape   : {df.shape}")
print(f"   Missing : {df.isnull().sum().sum()}")
print("📊 Ready for statistical analysis!")
```

```python
# ============================================
# CELL 5: STATISTICAL ANALYSIS Q1-Q6
# ============================================

print("=" * 45)
print("📊 STATISTICAL ANALYSIS")
print("=" * 45)

results = {}

# --- Q1: Active vs Inactive ---
active   = df[df['Customer Status']=='Active']['Total-Amt']
inactive = df[df['Customer Status']=='Inactive']['Total-Amt']
t_stat, p_value = stats.ttest_ind(active, inactive)
results['Q1'] = {
    'active_mean'  : active.mean(),
    'inactive_mean': inactive.mean(),
    'p_value'      : p_value,
    'significant'  : p_value < 0.05
}
print("✅ Q1 done: Active vs Inactive")

# --- Q2: Revenue by Product ---
results['Q2'] = {}
for col in amt_cols:
    name = col.replace('-Amt','')
    results['Q2'][name] = {
        'buyers'   : int((df[col] > 0).sum()),
        'total_rev': float(df[col].sum()),
        'avg_spend': float(df[df[col]>0][col].mean())
    }
print("✅ Q2 done: Revenue by Product")

# --- Q3: Buying Patterns ---
results['Q3'] = {
    'single_buyers': int((df['Products-Bought']==1).sum()),
    'multi_buyers' : int((df['Products-Bought']>1).sum()),
    'zero_buyers'  : int((df['Products-Bought']==0).sum()),
    'avg_products' : float(df['Products-Bought'].mean())
}
print("✅ Q3 done: Buying Patterns")

# --- Q4: Geographic ---
state_grp = df.groupby('state_cd').agg(
    Customers=('customer_id','count'),
    Total_Rev=('Total-Amt','sum'),
    Avg_Spend=('Total-Amt','mean')
).reset_index().sort_values('Total_Rev',
                             ascending=False)
results['Q4'] = {
    'top_state'    : state_grp.iloc[0]['state_cd'],
    'top_state_rev': float(state_grp.iloc[0]['Total_Rev']),
    'states_count' : int(df['state_cd'].nunique()),
    'cities_count' : int(df['city_nm'].nunique())
}
print("✅ Q4 done: Geographic Patterns")

# --- Q5: Customer Profile ---
results['Q5'] = {
    'mean_spend'  : float(df['Total-Amt'].mean()),
    'median_spend': float(df['Total-Amt'].median()),
    'top_spend'   : float(df['Total-Amt'].max()),
    'zero_spenders': int((df['Total-Amt']==0).sum())
}
print("✅ Q5 done: Customer Profile")

# --- Q6: Correlations ---
r_books_music, p_books_music = stats.pearsonr(
    df['Book-Amt'], df['Music CDs-Amt'])
results['Q6'] = {
    'books_music_r': float(r_books_music),
    'books_music_p': float(p_books_music),
    'significant'  : bool(p_books_music < 0.05)
}
print("✅ Q6 done: Correlations")

print("\n" + "=" * 45)
print("🎯 ALL 6 ANALYSES COMPLETE!")
print("=" * 45)
print("📊 Results ready for Magic Box!")
```

```python
# ============================================
# PHASE 4B: DATA CLEANING AGENT
# Handles messy data from ANY organization
# ============================================

def diagnose_data(dataframe):
    """
    Automatically diagnoses data quality
    Like a doctor checking patient health!
    """
    print("=" * 50)
    print("🔍 DATA CLEANING AGENT")
    print("   Diagnosing your data...")
    print("=" * 50)

    issues = []
    fixes  = []

    # Check 1: Shape
    rows, cols = dataframe.shape
    print(f"\n📐 DATASET SIZE:")
    print(f"   Rows    : {rows}")
    print(f"   Columns : {cols}")

    # Check 2: Missing values
    missing = dataframe.isnull().sum()
    total_missing = missing.sum()
    print(f"\n🔍 MISSING VALUES:")
    if total_missing == 0:
        print("   ✅ No missing values!")
    else:
        print(f"   ⚠️  Found {total_missing} missing values!")
        for col in missing[missing > 0].index:
            pct = missing[col]/rows*100
            print(f"   → {col}: {missing[col]} missing ({pct:.1f}%)")
            issues.append(f"Missing values in {col}")
            fixes.append(f"Fill {col} with 0 or median")

    # Check 3: Duplicate rows
    dupes = dataframe.duplicated().sum()
    print(f"\n🔍 DUPLICATE ROWS:")
    if dupes == 0:
        print("   ✅ No duplicates!")
    else:
        print(f"   ⚠️  Found {dupes} duplicate rows!")
        issues.append(f"{dupes} duplicate rows")
        fixes.append("Remove duplicate rows")

    # Check 4: Data types
    print(f"\n🔍 DATA TYPES:")
    for col in dataframe.columns:
        dtype = dataframe[col].dtype
        print(f"   {col:25s}: {str(dtype):10s}")

    # Check 5: Numeric columns check
    print(f"\n🔍 NUMERIC COLUMNS CHECK:")
    num_cols = dataframe.select_dtypes(
                include=[np.number]).columns
    for col in num_cols:
        zeros = (dataframe[col] == 0).sum()
        pct   = zeros/rows*100
        if pct > 90:
            print(f"   ⚠️  {col}: {pct:.1f}% zeros — check!")
            issues.append(f"{col} has {pct:.1f}% zeros")
        else:
            print(f"   ✅ {col}: looks good!")

    # Summary
    print(f"\n{'=' * 50}")
    print(f"📋 DIAGNOSIS SUMMARY:")
    print(f"{'=' * 50}")
    if len(issues) == 0:
        print("✅ Your data looks CLEAN!")
        print("🚀 Ready for DXPO analysis!")
    else:
        print(f"⚠️  Found {len(issues)} issues:")
        for i, (issue, fix) in enumerate(
                zip(issues, fixes), 1):
            print(f"\n   Issue {i}: {issue}")
            print(f"   Fix   : {fix}")
        print(f"\n💡 Shall Magic Box fix these?")
        response = input("Type YES to auto-fix: ")
        if response.upper() == "YES":
            print("\n🔧 Auto-fixing issues...")
            # Fix missing values
            for col in dataframe.columns:
                if dataframe[col].isnull().sum() > 0:
                    if dataframe[col].dtype in [
                            np.float64, np.int64]:
                        dataframe[col].fillna(0,
                            inplace=True)
                    else:
                        dataframe[col].fillna(
                            'Unknown', inplace=True)
            # Fix duplicates
            before = len(dataframe)
            dataframe.drop_duplicates(inplace=True)
            after = len(dataframe)
            if before != after:
                print(f"   ✅ Removed {before-after}"
                      f" duplicates!")
            print("   ✅ Missing values filled!")
            print("\n✅ Data cleaned successfully!")
            print("🚀 Ready for DXPO analysis!")
          else:
            print("\n✅ No auto-fix applied!")
            print("📊 Proceeding with original data!")
            print("🚀 Ready for DXPO analysis!")
    return dataframe

# Test it on our J-Bee data!
print("Testing Data Cleaning Agent")
print("on J-Bee dataset...\n")
df = diagnose_data(df)
```

```python
# ============================================
# 2D CUSTOMER VALUE MATRIX
# Decile Analysis — Pareto Style!
# By Dr. Jay Rajasekera's DXPO Framework
# ============================================

import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np
import pandas as pd
from datetime import date

today = date.today().strftime("%B %d, %Y")

print("=" * 50)
print("📊 BUILDING 2D CUSTOMER VALUE MATRIX...")
print("=" * 50)

# Step 1: Sort customers by spend
df_sorted = df.sort_values('Total-Amt',
                            ascending=False).reset_index(drop=True)

# Step 2: Assign deciles
df_sorted['Customer_Decile'] = pd.qcut(
    df_sorted.index + 1,
    q=10,
    labels=[f'Top\n{i*10}%' for i in range(1,11)])

df_sorted['Spend_Decile'] = pd.qcut(
    df_sorted['Total-Amt'].rank(
        method='first', ascending=False),
    q=10,
    labels=[f'Top\n{i*10}%' for i in range(1,11)])

# Step 3: Calculate matrix values
total_revenue = df_sorted['Total-Amt'].sum()
total_customers = len(df_sorted)

# Build 10x10 matrix
matrix_rev = np.zeros((10, 10))
matrix_cust = np.zeros((10, 10))

decile_labels = [f'Top {i*10}%'
                 for i in range(1, 11)]

for i in range(10):
    # Customer decile boundaries
    cust_start = i * (total_customers // 10)
    cust_end = (i+1) * (total_customers // 10)
    cust_group = df_sorted.iloc[cust_start:cust_end]

    for j in range(10):
        # Spend decile boundaries
        spend_start = j * (total_customers // 10)
        spend_end = (j+1) * (total_customers // 10)
        spend_threshold_high = df_sorted.iloc[
            spend_start]['Total-Amt']
        spend_threshold_low = df_sorted.iloc[
            min(spend_end,
                total_customers-1)]['Total-Amt']

        # Customers in this cell
        cell_customers = cust_group[
            (cust_group['Total-Amt'] <=
             spend_threshold_high) &
            (cust_group['Total-Amt'] >=
             spend_threshold_low)]

        matrix_cust[j][i] = len(cell_customers)
        matrix_rev[j][i] = cell_customers[
            'Total-Amt'].sum()

# Step 4: Create the visualization
fig, axes = plt.subplots(1, 2,
                          figsize=(20, 10))
fig.patch.set_facecolor('#F8F9FA')

# --- Plot 1: Customer Count Matrix ---
ax1 = axes[0]
im1 = ax1.imshow(matrix_cust,
                  cmap='YlOrRd',
                  aspect='auto')

# Add text to cells
for i in range(10):
    for j in range(10):
        val = int(matrix_cust[i][j])
        if val > 0:
            ax1.text(j, i, str(val),
                    ha='center',
                    va='center',
                    fontsize=9,
                    fontweight='bold',
                    color='black' if val < 30
                    else 'white')

ax1.set_xticks(range(10))
ax1.set_yticks(range(10))
ax1.set_xticklabels(
    [f'Top\n{(i+1)*10}%' for i in range(10)],
    fontsize=8)
ax1.set_yticklabels(
    [f'Top\n{(i+1)*10}%' for i in range(10)],
    fontsize=8)
ax1.set_xlabel(
    'Customer Range\n(Top Spenders → Left)',
    fontsize=12, fontweight='bold')
ax1.set_ylabel(
    'Spend Range\n(Top Spenders → Top)',
    fontsize=12, fontweight='bold')
ax1.set_title(
    '👥 Customer Count Matrix\n'
    'by Spend & Customer Deciles',
    fontsize=14, fontweight='bold',
    color='#1B3A6B', pad=15)
plt.colorbar(im1, ax=ax1,
             label='Number of Customers')

# --- Plot 2: Revenue Matrix ---
ax2 = axes[1]
im2 = ax2.imshow(matrix_rev,
                  cmap='YlOrRd',
                  aspect='auto')

# Add text to cells
for i in range(10):
    for j in range(10):
        val = matrix_rev[i][j]
        if val > 0:
            ax2.text(j, i,
                    f'${val:,.0f}',
                    ha='center',
                    va='center',
                    fontsize=7,
                    fontweight='bold',
                    color='black' if val < 5000
                    else 'white')

ax2.set_xticks(range(10))
ax2.set_yticks(range(10))
ax2.set_xticklabels(
    [f'Top\n{(i+1)*10}%' for i in range(10)],
    fontsize=8)
ax2.set_yticklabels(
    [f'Top\n{(i+1)*10}%' for i in range(10)],
    fontsize=8)
ax2.set_xlabel(
    'Customer Range\n(Top Spenders → Left)',
    fontsize=12, fontweight='bold')
ax2.set_ylabel(
    'Spend Range\n(Top Spenders → Top)',
    fontsize=12, fontweight='bold')
ax2.set_title(
    '💰 Revenue Matrix\n'
    'by Spend & Customer Deciles',
    fontsize=14, fontweight='bold',
    color='#1B3A6B', pad=15)
plt.colorbar(im2, ax=ax2,
             label='Revenue ($)')

# Main title
fig.suptitle(
    '🎯 DXPO Customer Value Matrix  |  '
    f'J-Bee Analysis  |  {today}\n'
    'Dr. Jay Rajasekera — '
    'Tokyo International University',
    fontsize=14,
    fontweight='bold',
    color='#1B3A6B',
    y=1.02)

plt.tight_layout()

# Save and download
plt.savefig('DXPO_Customer_Matrix.png',
            dpi=150,
            bbox_inches='tight',
            facecolor='#F8F9FA')
plt.show()

# Pareto Analysis
print("\n" + "=" * 50)
print("📊 PARETO ANALYSIS SUMMARY")
print("=" * 50)

cumrev = 0
cumcust = 0
print(f"\n{'Decile':12s} {'Customers':>10s} "
      f"{'Revenue':>12s} {'Cum Rev%':>10s}")
print("-" * 46)

for i in range(10):
    cust_start = i * (total_customers // 10)
    cust_end = (i+1) * (total_customers // 10)
    grp = df_sorted.iloc[cust_start:cust_end]
    rev = grp['Total-Amt'].sum()
    cumrev += rev
    cumcust += len(grp)
    pct = cumrev/total_revenue*100
    print(f"Top {(i+1)*10:2d}%     "
          f"{cumcust:>10d} "
          f"${cumrev:>11,.2f} "
          f"{pct:>9.1f}%")

print("\n" + "=" * 50)
print("💾 Saving matrix for download...")
from google.colab import files
files.download('DXPO_Customer_Matrix.png')
print("📥 Download started!")
print("🎨 Ready for PPT!!")
```

```python
# ============================================
# MARKETING DOKKU v2
# Adaptive tests based on DXPO DOC v2 context
# Primary + Secondary answers!!
# ============================================

print("=" * 55)
print("🔬 MARKETING DOKKU v2 — Adaptive Tests")
print(f"   PRIMARY concern : {doc_context['concern_primary']}")
if doc_context['concern_secondary']:
    print(f"   SECONDARY concern: {doc_context['concern_secondary']}")
print(f"   PRIMARY dept    : {doc_context['dept_primary']}")
if doc_context['dept_secondary']:
    print(f"   SECONDARY dept  : {doc_context['dept_secondary']}")
print("=" * 55)

amt_cols = ['Book-Amt','Music CDs-Amt',
            'Housewares-Amt','Tools-Amt',
            'Games-Amt','Gift-Amt',
            'Video-Amt','Magazine-Amt',
            'CellPhoneService-Amt',
            'MiscProduct-Amt']

channel_cols = [
    'Internet Order Frequency',
    'Fax Order Frequency',
    'Direct Mail Order  Frequency',
    'Phone Order Frequency',
    'Point Of Sales Order Frequency'
]

df['Total-Amt'] = df[amt_cols].sum(axis=1)
df['Products-Bought'] = (
    df[amt_cols] > 0).sum(axis=1)

n = len(df)
pain_points = []

# ══════════════════════════════════════════
# STANDARD TESTS — always run (5 tests)
# ══════════════════════════════════════════

print("\n📋 STANDARD TEST BATTERY")
print("─" * 55)

# ── STANDARD TEST 1: Customer Segmentation ──
print("\n🔬 S-TEST 1: Customer Segmentation")
mean_spend   = df['Total-Amt'].mean()
median_spend = df['Total-Amt'].median()
zero_spend   = (df['Total-Amt']==0).sum()
zero_pct     = zero_spend/n*100

print(f"  Mean spend   : ${mean_spend:,.2f}")
print(f"  Median spend : ${median_spend:,.2f}")
print(f"  Zero spenders: {zero_spend:,} ({zero_pct:.1f}%)")

if zero_pct > 20:
    flag = "🔴 RED"; sev = 3
    pain_points.append({
        "test"    : "Customer Segmentation",
        "finding" : f"{zero_pct:.1f}% never spent",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
elif zero_pct > 10:
    flag = "🟡 YELLOW"; sev = 2
    pain_points.append({
        "test"    : "Customer Segmentation",
        "finding" : f"{zero_pct:.1f}% never spent",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
else:
    flag = "🟢 GREEN"
print(f"  Status       : {flag}")

# ── STANDARD TEST 2: Retention ──
print("\n🔬 S-TEST 2: Acquisition vs Retention")
active   = (df['Customer Status']=='Active').sum()
inactive = (df['Customer Status']=='Inactive').sum()
inactive_pct = inactive/n*100
active_spend = df[
    df['Customer Status']=='Active'][
    'Total-Amt'].mean()
inactive_spend = df[
    df['Customer Status']=='Inactive'][
    'Total-Amt'].mean()
spend_gap = active_spend - inactive_spend

print(f"  Active   : {active:,} ({100-inactive_pct:.1f}%)"
      f" · avg ${active_spend:,.2f}")
print(f"  Inactive : {inactive:,} ({inactive_pct:.1f}%)"
      f" · avg ${inactive_spend:,.2f}")
print(f"  Spend gap: ${spend_gap:,.2f} per customer")

if inactive_pct > 30:
    flag = "🔴 RED"; sev = 3
    pain_points.append({
        "test"    : "Customer Retention",
        "finding" : f"{inactive_pct:.1f}% inactive"
                    f" · ${spend_gap:,.0f} gap",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
elif inactive_pct > 15:
    flag = "🟡 YELLOW"; sev = 2
    pain_points.append({
        "test"    : "Customer Retention",
        "finding" : f"{inactive_pct:.1f}% inactive"
                    f" · ${spend_gap:,.0f} gap",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
else:
    flag = "🟢 GREEN"
print(f"  Status   : {flag}")

# ── STANDARD TEST 3: Cross-sell ──
print("\n🔬 S-TEST 3: Cross-sell Penetration")
single_pct = (df['Products-Bought']==1).sum()/n*100
multi_pct  = (df['Products-Bought']>1).sum()/n*100

print(f"  Single product: {single_pct:.1f}%")
print(f"  Multi product : {multi_pct:.1f}%")

if single_pct > 60:
    flag = "🔴 RED"; sev = 3
    pain_points.append({
        "test"    : "Cross-sell Penetration",
        "finding" : f"{single_pct:.1f}% single buyers",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
elif single_pct > 40:
    flag = "🟡 YELLOW"; sev = 2
    pain_points.append({
        "test"    : "Cross-sell Penetration",
        "finding" : f"{single_pct:.1f}% single buyers",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
else:
    flag = "🟢 GREEN"
print(f"  Status        : {flag}")

# ── STANDARD TEST 4: Digital Channel Gap ──
print("\n🔬 S-TEST 4: Digital Channel Gap")
internet_pct = (
    df['Internet Order Frequency']>0
    ).sum()/n*100
mail_pct = (
    df['Direct Mail Order  Frequency']>0
    ).sum()/n*100

print(f"  Internet orders: {internet_pct:.1f}%")
print(f"  Direct Mail    : {mail_pct:.1f}%")
print(f"  Digital gap    : {mail_pct-internet_pct:.1f}%")

if internet_pct < 10:
    flag = "🔴 RED"; sev = 3
    pain_points.append({
        "test"    : "Digital Channel Gap",
        "finding" : f"Only {internet_pct:.1f}%"
                    f" ordering online",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
elif internet_pct < 25:
    flag = "🟡 YELLOW"; sev = 2
    pain_points.append({
        "test"    : "Digital Channel Gap",
        "finding" : f"Only {internet_pct:.1f}%"
                    f" ordering online",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
else:
    flag = "🟢 GREEN"
print(f"  Status         : {flag}")

# ── STANDARD TEST 5: Revenue Concentration ──
print("\n🔬 S-TEST 5: Revenue Concentration")
rev_by_prod = {
    col.replace('-Amt',''): df[col].sum()
    for col in amt_cols}
total_rev   = sum(rev_by_prod.values())
top_prod    = max(rev_by_prod,
                  key=rev_by_prod.get)
top_pct     = rev_by_prod[top_prod]/total_rev*100

print(f"  Top product  : {top_prod}")
print(f"  Revenue share: {top_pct:.1f}%")

if top_pct > 40:
    flag = "🔴 RED"; sev = 3
    pain_points.append({
        "test"    : "Revenue Concentration",
        "finding" : f"{top_prod} = {top_pct:.1f}%"
                    f" of revenue",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
elif top_pct > 25:
    flag = "🟡 YELLOW"; sev = 2
    pain_points.append({
        "test"    : "Revenue Concentration",
        "finding" : f"{top_prod} = {top_pct:.1f}%"
                    f" of revenue",
        "severity": sev, "flag": flag,
        "type"    : "standard"})
else:
    flag = "🟢 GREEN"
print(f"  Status       : {flag}")

# ══════════════════════════════════════════
# ADAPTIVE TESTS — triggered by DOC answers
# ══════════════════════════════════════════

adaptive_tests_run = []

print("\n\n📋 ADAPTIVE TEST BATTERY")
print("   (triggered by YOUR DOC answers)")
print("─" * 55)

# ── ADAPTIVE: Customer Churn ──
churn_triggers = [
    "Customer churn",
    "Customer identification",
    "Marketing effectiveness"
]
if (doc_context['concern_primary'] in churn_triggers or
    doc_context.get('concern_secondary')
    in churn_triggers):

    print("\n🔬 A-TEST: Churn Risk Analysis")
    print("   (triggered by your concern answers)")
    adaptive_tests_run.append("Churn Risk")

    # Churn indicators
    inactive_spend_pct = (
        inactive_spend/active_spend*100)
    revenue_at_risk = inactive * inactive_spend

    print(f"  Inactive spend vs Active: "
          f"{inactive_spend_pct:.1f}%")
    print(f"  Revenue at risk if churned: "
          f"${revenue_at_risk:,.2f}")
    print(f"  Recovery potential if "
          f"reactivated 50%: "
          f"${revenue_at_risk*0.5:,.2f}")

    if inactive_spend_pct < 60:
        flag = "🔴 RED"; sev = 3
        pain_points.append({
            "test"    : "Churn Risk",
            "finding" : f"Inactive spend only"
                        f" {inactive_spend_pct:.1f}%"
                        f" of active · "
                        f"${revenue_at_risk:,.0f}"
                        f" at risk",
            "severity": sev, "flag": flag,
            "type"    : "adaptive"})
    elif inactive_spend_pct < 80:
        flag = "🟡 YELLOW"; sev = 2
        pain_points.append({
            "test"    : "Churn Risk",
            "finding" : f"Inactive spend"
                        f" {inactive_spend_pct:.1f}%"
                        f" of active",
            "severity": sev, "flag": flag,
            "type"    : "adaptive"})
    else:
        flag = "🟢 GREEN"
    print(f"  Status: {flag}")

# ── ADAPTIVE: Customer Service ──
service_triggers = [
    "Customer Service",
    "Sales & Marketing"
]
if (doc_context['dept_primary'] in service_triggers or
    doc_context.get('dept_secondary')
    in service_triggers):

    print("\n🔬 A-TEST: Customer Value Distribution")
    print("   (triggered by your department answers)")
    adaptive_tests_run.append(
        "Customer Value Distribution")

    # Value segments
    top10_threshold = df['Total-Amt'].quantile(0.90)
    top10_customers = df[
        df['Total-Amt'] >= top10_threshold]
    top10_revenue   = top10_customers[
        'Total-Amt'].sum()
    top10_pct_rev   = top10_revenue/total_rev*100

    print(f"  Top 10% customers spend "
          f">${top10_threshold:,.0f}+")
    print(f"  Top 10% generate "
          f"{top10_pct_rev:.1f}% of revenue")
    print(f"  Bottom 50% generate "
          f"{100-top10_pct_rev:.1f}% of revenue")

    if top10_pct_rev > 50:
        flag = "🔴 RED"; sev = 3
        pain_points.append({
            "test"    : "Customer Value Distribution",
            "finding" : f"Top 10% customers = "
                        f"{top10_pct_rev:.1f}%"
                        f" of revenue · "
                        f"high dependency risk",
            "severity": sev, "flag": flag,
            "type"    : "adaptive"})
    elif top10_pct_rev > 35:
        flag = "🟡 YELLOW"; sev = 2
        pain_points.append({
            "test"    : "Customer Value Distribution",
            "finding" : f"Top 10% = "
                        f"{top10_pct_rev:.1f}%"
                        f" of revenue",
            "severity": sev, "flag": flag,
            "type"    : "adaptive"})
    else:
        flag = "🟢 GREEN"
    print(f"  Status: {flag}")

# ── ADAPTIVE: Marketing Effectiveness ──
if (doc_context['concern_primary'] ==
    "Marketing effectiveness" or
    doc_context.get('concern_secondary') ==
    "Marketing effectiveness"):

    print("\n🔬 A-TEST: Product Mix Effectiveness")
    print("   (triggered by marketing concern)")
    adaptive_tests_run.append(
        "Product Mix Effectiveness")

    # Which products have low penetration
    low_penetration = []
    for col in amt_cols:
        prod = col.replace('-Amt','')
        buyers = (df[col]>0).sum()
        pct = buyers/n*100
        if pct < 15:
            low_penetration.append(
                f"{prod} ({pct:.1f}%)")

    print(f"  Low penetration products "
          f"(<15% customers):")
    if low_penetration:
        for p in low_penetration:
            print(f"    → {p}")
        flag = "🟡 YELLOW"; sev = 2
        pain_points.append({
            "test"    : "Product Mix",
            "finding" : f"{len(low_penetration)}"
                        f" products under 15%"
                        f" penetration",
            "severity": sev, "flag": flag,
            "type"    : "adaptive"})
    else:
        print("    ✅ All products healthy!")
        flag = "🟢 GREEN"
    print(f"  Status: {flag}")

# ══════════════════════════════════════════
# FULL DIAGNOSIS SUMMARY
# ══════════════════════════════════════════

print("\n\n" + "=" * 55)
print("🏥 MARKETING DOKKU v2 — FULL DIAGNOSIS")
print("=" * 55)

std_points = [p for p in pain_points
              if p['type']=='standard']
adp_points = [p for p in pain_points
              if p['type']=='adaptive']
red_points = [p for p in pain_points
              if p['flag']=='🔴 RED']
yel_points = [p for p in pain_points
              if p['flag']=='🟡 YELLOW']

print(f"\n  Standard tests run : 5")
print(f"  Adaptive tests run : "
      f"{len(adaptive_tests_run)}")
print(f"  Total tests run    : "
      f"{5+len(adaptive_tests_run)}")
print(f"\n  🔴 RED flags   : {len(red_points)}")
for p in red_points:
    t = "STD" if p['type']=='standard' else "ADP"
    print(f"     [{t}] {p['test']}: {p['finding']}")

print(f"\n  🟡 YELLOW flags: {len(yel_points)}")
for p in yel_points:
    t = "STD" if p['type']=='standard' else "ADP"
    print(f"     [{t}] {p['test']}: {p['finding']}")

print(f"\n  Total pain points  : {len(pain_points)}")
print(f"\n✅ Standard tests always run!!")
print(f"✅ Adaptive tests triggered by")
print(f"   YOUR specific DOC answers!!")
print("\n🎯 Ready for Strategic Ranking (Obj-2)!!")
print("=" * 55)
```

```python
# ============================================
# DXPO IMPACT EVALUATION
# Based on Dr. Jay Rajasekera's
# APO Framework + Springer Book!!
# Figure 4 — Impact = A × B
# ============================================

print("=" * 55)
print("🎯 DXPO IMPACT EVALUATION PROCESS")
print("   Based on Dr. Jay Rajasekera")
print("   APO Framework · Springer 2024")
print("=" * 55)
print("""
For each Pain Point identified
by Marketing Dokku:

  A: Rate CUSTOMER VALUE (1-10)
     → How valuable is solving this?
     → Impact on customers/revenue?

  B: Rate IMPLEMENTATION POTENTIAL (1-10)
     → Does J-Bee have resources?
     → Internal capability available?
     → 10 = fully capable internally
     →  1 = needs full external help

  IMPACT = A × B
  MAX possible = 100!!
""")

print("─" * 55)
print("Pain points found by Marketing Dokku:")
print("─" * 55)

for i, p in enumerate(pain_points, 1):
    print(f"  {i}. {p['test']}")
    print(f"     Finding: {p['finding']}")
    print(f"     Flag   : {p['flag']}")
    print()

print("─" * 55)
print("Please score each pain point:")
print("─" * 55)

impact_scores = []

for i, p in enumerate(pain_points, 1):
    print(f"\n{'─'*40}")
    print(f"Pain Point {i}: {p['test']}")
    print(f"Finding: {p['finding']}")
    print(f"Flag   : {p['flag']}")
    print()

    # Get Customer Value score
    while True:
        try:
            a = int(input(
                f"  A: Customer Value (1-10): "))
            if 1 <= a <= 10:
                break
            print("  ⚠️  Please enter 1-10!")
        except:
            print("  ⚠️  Please enter a number!")

    # Get Implementation Potential
    while True:
        try:
            b = int(input(
                f"  B: Implementation "
                f"Potential (1-10): "))
            if 1 <= b <= 10:
                break
            print("  ⚠️  Please enter 1-10!")
        except:
            print("  ⚠️  Please enter a number!")

    impact = a * b
    impact_scores.append({
        "pain_point"  : p['test'],
        "finding"     : p['finding'],
        "flag"        : p['flag'],
        "customer_value"      : a,
        "implementation_potential" : b,
        "impact"      : impact
    })
    print(f"  ✅ Impact = {a} × {b} = {impact}")

# Sort by Impact
impact_scores.sort(
    key=lambda x: x['impact'],
    reverse=True)

# Assign ranks
for i, p in enumerate(impact_scores, 1):
    p['rank'] = i

# ── DISPLAY FIGURE 4 STYLE TABLE ──
print("\n\n" + "=" * 55)
print("📊 DXPO IMPACT EVALUATION TABLE")
print("   (Figure 4 style · Dr. Jay Rajasekera)")
print("=" * 55)

print(f"\n{'Process':<28} {'A':>4} {'B':>4} "
      f"{'Impact':>8} {'Rank':>6}")
print("─" * 55)

for p in impact_scores:
    rank_color = "🥇" if p['rank']==1 else \
                 "🥈" if p['rank']==2 else \
                 "🥉" if p['rank']==3 else \
                 f"#{p['rank']}"
    print(f"{p['pain_point']:<28} "
          f"{p['customer_value']:>4} "
          f"{p['implementation_potential']:>4} "
          f"{p['impact']:>8} "
          f"{rank_color:>6}")

print("─" * 55)
print(f"\n  MAX possible Impact = 10 × 10 = 100")
print(f"\n  🥇 TOP PRIORITY:")
print(f"     {impact_scores[0]['pain_point']}")
print(f"     Impact = {impact_scores[0]['impact']}")
print(f"     → FIX THIS FIRST!!")

print(f"\n  ⚡ QUICK WIN:")
print(f"     {impact_scores[1]['pain_point']}")
print(f"     Impact = {impact_scores[1]['impact']}")

print("\n" + "=" * 55)
print("✅ DXPO Impact Evaluation Complete!!")
print("🎯 Ranking based on YOUR framework!!")
print("=" * 55)
```
